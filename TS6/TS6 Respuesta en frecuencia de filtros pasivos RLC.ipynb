#### Análisis y Procesamiento de Señales

# TS6: Respuesta en frecuencia de filtros pasivos RLC

## Introducción

En este trabajo se analiza la respuesta en frecuencia de tres filtros pasivos RLC (circuitos a, b y c) y del circuito del bonus. Para cada uno se pide:

1. Hallar analíticamente la función de transferencia $H(s) = \frac{V_2(s)}{V_1(s)}$ para $\omega_0 = 1$, $R = 1\ \Omega$ y $Q = \frac{\sqrt{2}}{2}$.
2. Calcular y graficar la respuesta de módulo y fase, indicando la respuesta asintótica.
3. Determinar qué tipo de filtro es y cuál es su ancho de banda.

En el bonus además se modifican los valores de los componentes para que el circuito opere en $f_0 = 10\ \text{kHz}$ con $R = 50\ \Omega$.


## Desarrollo

### Circuito a)



#### a.1) Función de transferencia


Impedancias:

$$Z_L = sL \qquad Z_R = R \qquad Z_C = \frac{1}{sC}$$

Impedancia vista desde la entrada (los tres en serie):

$$Z_{eq} = sL + R + \frac{1}{sC} = \frac{s^2LC + sRC + 1}{sC}$$

Ecuaciones de malla:

$$V_1 = I_1\, Z_{eq} = I_1\left(\frac{s^2LC + sRC + 1}{sC}\right) \qquad (1)$$

$$V_2 = I_1\, Z_C = I_1\left(\frac{1}{sC}\right) \qquad (2)$$

Despejo $I_1$ de (2):

$$I_1 = V_2\,(sC) \qquad (3)$$

Reemplazo (3) en (1):

$$V_1 = V_2\,(sC)\left(\frac{s^2LC + sRC + 1}{sC}\right) = V_2\left(s^2LC + sRC + 1\right) \qquad (4)$$

Entonces:

$$H(s) = \frac{V_2}{V_1} = \frac{1}{s^2LC + sRC + 1}$$

Divido numerador y denominador por $LC$:

$$H(s) = \frac{\frac{1}{LC}}{s^2 + s\,\frac{R}{L} + \frac{1}{LC}} \qquad (5)$$

Comparando con la forma normalizada de 2.º orden:

$$\omega_0^2 = \frac{1}{LC} \qquad (6) \qquad\qquad \frac{\omega_0}{Q} = \frac{R}{L} \;\Rightarrow\; Q = \frac{\omega_0 L}{R} \qquad (7)$$

Reemplazo (6) y (7) en (5):

$$H(s) = \frac{\omega_0^2}{s^2 + s\,\frac{\omega_0}{Q} + \omega_0^2}$$

Con $\omega_0 = 1$, $R = 1\,\Omega$ y $Q = \frac{\sqrt{2}}{2}$ queda $\frac{\omega_0}{Q} = \sqrt{2}$:

$$\boxed{H(s) = \frac{1}{s^2 + \sqrt{2}\,s + 1}}$$

Y los componentes:

$$L = \frac{Q\,R}{\omega_0} = \frac{\sqrt{2}}{2}\ \text{H} \qquad C = \frac{1}{\omega_0^2\,L} = \sqrt{2}\ \text{F}$$

#### a.2) Respuesta de módulo y fase

Reemplazo $s = j\omega$:

$$H(j\omega) = \frac{1}{(1 - \omega^2) + j\sqrt{2}\,\omega}$$

Módulo:

$$|H(j\omega)| = \frac{1}{\sqrt{(1-\omega^2)^2 + 2\omega^2}} = \frac{1}{\sqrt{1 + \omega^4}} \qquad\Rightarrow\qquad |H(j\omega)|_{dB} = -10\,\log_{10}\left(1 + \omega^4\right)$$

Fase:

$$\varphi(\omega) = -\arctan\left(\frac{\sqrt{2}\,\omega}{1 - \omega^2}\right)$$

(para $\omega > 1$ el denominador es negativo y el ángulo cae en el segundo cuadrante).

Asíntotas:

- $\omega \ll \omega_0$: $|H| \approx 1 \Rightarrow 0\ \text{dB}$, $\varphi \to 0°$
- $\omega \gg \omega_0$: $|H| \approx \dfrac{1}{\omega^2} \Rightarrow -40\ \text{dB/déc}$, $\varphi \to -180°$
- En $\omega = \omega_0$: $|H| = \dfrac{1}{\sqrt{2}} \Rightarrow -3{,}01\ \text{dB}$, $\varphi = -90°$


### Circuito b)



#### b.1) Función de transferencia



Impedancias:

$$Z_L = sL \qquad Z_R = R \qquad Z_C = \frac{1}{sC}$$

Paralelo $R \parallel C$:

$$Z_{RC} = \frac{R \cdot \frac{1}{sC}}{R + \frac{1}{sC}} = \frac{R}{sRC + 1}$$

Impedancia vista desde la entrada ($Z_{RC}$ en serie con $L$):

$$Z_{eq} = \frac{R}{sRC + 1} + sL = \frac{s^2RLC + sL + R}{sRC + 1}$$

Ecuaciones de malla:

$$V_1 = I_1\, Z_{eq} = I_1\left(\frac{s^2RLC + sL + R}{sRC + 1}\right) \qquad (1)$$

$$V_2 = I_1\, Z_L = I_1\,(sL) \qquad (2)$$

Despejo $I_1$ de (2):

$$I_1 = V_2\left(\frac{1}{sL}\right) \qquad (3)$$

Reemplazo (3) en (1):

$$V_1 = V_2\left(\frac{1}{sL}\right)\left(\frac{s^2RLC + sL + R}{sRC + 1}\right) = V_2\left(\frac{s^2RLC + sL + R}{s^2RLC + sL}\right) \qquad (4)$$

Entonces:

$$H(s) = \frac{V_2}{V_1} = \frac{s^2RLC + sL}{s^2RLC + sL + R}$$

Divido numerador y denominador por $RLC$:

$$H(s) = \frac{s^2 + s\,\frac{1}{RC}}{s^2 + s\,\frac{1}{RC} + \frac{1}{LC}} \qquad (5)$$

Comparando con la forma normalizada:

$$\omega_0^2 = \frac{1}{LC} \qquad (6) \qquad\qquad \frac{\omega_0}{Q} = \frac{1}{RC} \;\Rightarrow\; Q = R\,\omega_0 C \qquad (7)$$

Reemplazo (6) y (7) en (5):

$$H(s) = \frac{s^2 + s\,\frac{\omega_0}{Q}}{s^2 + s\,\frac{\omega_0}{Q} + \omega_0^2}$$

Con $\omega_0 = 1$, $R = 1\,\Omega$ y $Q = \frac{\sqrt{2}}{2}$ queda $\frac{\omega_0}{Q} = \sqrt{2}$:

$$\boxed{H(s) = \frac{s^2 + \sqrt{2}\,s}{s^2 + \sqrt{2}\,s + 1} = \frac{s\left(s + \sqrt{2}\right)}{s^2 + \sqrt{2}\,s + 1}}$$

Y los componentes:

$$C = \frac{Q}{\omega_0\,R} = \frac{\sqrt{2}}{2}\ \text{F} \qquad L = \frac{1}{\omega_0^2\,C} = \sqrt{2}\ \text{H}$$

#### b.2) Respuesta de módulo y fase

$H(s)$ tiene un cero en el origen, un cero real en $\omega_z = \sqrt{2}$ y el par de polos complejos con $\omega_0 = 1$, $Q = \frac{\sqrt{2}}{2}$.

Reemplazo $s = j\omega$:

$$H(j\omega) = \frac{j\omega\left(\sqrt{2} + j\omega\right)}{(1 - \omega^2) + j\sqrt{2}\,\omega}$$

Módulo:

$$|H(j\omega)| = \frac{\omega\,\sqrt{2 + \omega^2}}{\sqrt{(1-\omega^2)^2 + 2\omega^2}} = \frac{\omega\,\sqrt{2 + \omega^2}}{\sqrt{1 + \omega^4}}$$

Fase:

$$\varphi(\omega) = 90° + \arctan\left(\frac{\omega}{\sqrt{2}}\right) - \arctan\left(\frac{\sqrt{2}\,\omega}{1 - \omega^2}\right)$$

(en el último término, para $\omega > 1$ se toma el ángulo del segundo cuadrante).

Asíntotas de módulo:

- $\omega < \omega_0$: $|H| \approx \sqrt{2}\,\omega \Rightarrow +20\ \text{dB/déc}$ (cero en el origen)
- $\omega_0 < \omega < \omega_z$: el polo doble suma $-40\ \text{dB/déc}$, la pendiente queda en $-20\ \text{dB/déc}$
- $\omega > \omega_z$: el cero en $\sqrt{2}$ suma $+20\ \text{dB/déc}$, la pendiente queda en $0$ y $|H| \to 1$ ($0\ \text{dB}$)


$\varphi \to +90°$ en bajas frecuencias y $\varphi \to 0°$ en altas.

### Circuito c)



#### c.1) Función de transferencia


Impedancias:

$$Z_L = sL \qquad Z_R = R \qquad Z_C = \frac{1}{sC}$$

Paralelo $L \parallel C$:

$$Z_{LC} = \frac{sL \cdot \frac{1}{sC}}{sL + \frac{1}{sC}} = \frac{sL}{s^2LC + 1}$$

Impedancia vista desde la entrada ($R$ en serie con $Z_{LC}$):

$$Z_{eq} = R + \frac{sL}{s^2LC + 1} = \frac{s^2RLC + sL + R}{s^2LC + 1}$$

Ecuaciones de malla:

$$V_1 = I_1\, Z_{eq} = I_1\left(\frac{s^2RLC + sL + R}{s^2LC + 1}\right) \qquad (1)$$

$$V_2 = I_1\, Z_{LC} = I_1\left(\frac{sL}{s^2LC + 1}\right) \qquad (2)$$

Despejo $I_1$ de (2):

$$I_1 = V_2\left(\frac{s^2LC + 1}{sL}\right) \qquad (3)$$

Reemplazo (3) en (1), el factor $(s^2LC + 1)$ se simplifica:

$$V_1 = V_2\left(\frac{s^2LC + 1}{sL}\right)\left(\frac{s^2RLC + sL + R}{s^2LC + 1}\right) = V_2\left(\frac{s^2RLC + sL + R}{sL}\right) \qquad (4)$$

Entonces:

$$H(s) = \frac{V_2}{V_1} = \frac{sL}{s^2RLC + sL + R}$$

Divido numerador y denominador por $RLC$:

$$H(s) = \frac{s\,\frac{1}{RC}}{s^2 + s\,\frac{1}{RC} + \frac{1}{LC}} \qquad (5)$$

Comparando con la forma normalizada:

$$\omega_0^2 = \frac{1}{LC} \qquad (6) \qquad\qquad \frac{\omega_0}{Q} = \frac{1}{RC} \;\Rightarrow\; Q = R\,\omega_0 C \qquad (7)$$

Reemplazo (6) y (7) en (5):

$$H(s) = \frac{s\,\frac{\omega_0}{Q}}{s^2 + s\,\frac{\omega_0}{Q} + \omega_0^2}$$

Con $\omega_0 = 1$, $R = 1\,\Omega$ y $Q = \frac{\sqrt{2}}{2}$ queda $\frac{\omega_0}{Q} = \sqrt{2}$:

$$\boxed{H(s) = \frac{\sqrt{2}\,s}{s^2 + \sqrt{2}\,s + 1}}$$

Y los componentes:

$$C = \frac{Q}{\omega_0\,R} = \frac{\sqrt{2}}{2}\ \text{F} \qquad L = \frac{1}{\omega_0^2\,C} = \sqrt{2}\ \text{H}$$

#### c.2) Respuesta de módulo y fase

Reemplazo $s = j\omega$:

$$H(j\omega) = \frac{j\sqrt{2}\,\omega}{(1 - \omega^2) + j\sqrt{2}\,\omega}$$

Módulo:

$$|H(j\omega)| = \frac{\sqrt{2}\,\omega}{\sqrt{(1-\omega^2)^2 + 2\omega^2}} = \frac{\sqrt{2}\,\omega}{\sqrt{1 + \omega^4}}$$

Fase:

$$\varphi(\omega) = 90° - \arctan\left(\frac{\sqrt{2}\,\omega}{1 - \omega^2}\right)$$

(para $\omega > 1$ se toma el ángulo del segundo cuadrante).

Asíntotas:

- $\omega \ll \omega_0$: $|H| \approx \sqrt{2}\,\omega \Rightarrow +20\ \text{dB/déc}$, $\varphi \to +90°$
- $\omega \gg \omega_0$: $|H| \approx \dfrac{\sqrt{2}}{\omega} \Rightarrow -20\ \text{dB/déc}$, $\varphi \to -90°$
- En $\omega = \omega_0$: $|H| = 1 \Rightarrow 0\ \text{dB}$, $\varphi = 0°$


### Bonus



#### bonus.1) Función de transferencia

Admitancias del circuito:

- Rama serie: $s\frac{L}{b}$ en paralelo con $\frac{1}{sCa}$
- Rama paralelo: $\frac{1}{sC(1-a)}$, $s\frac{L}{1-b}$ y $R$

$$Y_1 = \frac{b}{sL} + sCa \qquad (1)$$

$$Y_2 = sC - sCa + \frac{1}{sL} - \frac{b}{sL} + \frac{1}{R} \qquad (2)$$

Como la salida está en vacío ($I_2 = 0$), por $Z_1$ y $Z_2$ circula la misma corriente $I_1$:

$$V_1 = I_1(Z_1 + Z_2) = I_1\left(\frac{1}{Y_1} + \frac{1}{Y_2}\right) = I_1\left(\frac{Y_1 + Y_2}{Y_1 Y_2}\right) \qquad (3)$$

$$V_2 = I_1 Z_2 = I_1\left(\frac{1}{Y_2}\right) \qquad (4)$$

Despejando de (4):

$$I_1 = V_2 \cdot Y_2 \qquad (5)$$

Reemplazando (5) en (3):

$$V_1 = V_2\, Y_2 \left(\frac{Y_1 + Y_2}{Y_1 Y_2}\right) = V_2\left(\frac{Y_1+Y_2}{Y_1}\right) \qquad (6)$$

Reemplazando (1) y (2) en (6):

$$V_1 = V_2\left(\frac{sL}{b + s^2LCa}\right)\left[\frac{b}{sL} + sCa + sC - sCa + \frac{1}{sL} - \frac{b}{sL} + \frac{1}{R}\right]$$

$$V_1 = V_2\left(\frac{sL}{b + s^2LCa}\right)\left(sC + \frac{1}{sL} + \frac{1}{R}\right)$$

$$V_1 = V_2\left(\frac{sL}{b + s^2LCa}\right)\left(\frac{s^2LCR + sL + R}{sLR}\right)$$

$$V_1 = V_2\left(\frac{s^2LCR + sL + R}{bR + s^2LCRa}\right)$$

Por lo tanto, la función de transferencia $H(s) = \dfrac{V_2}{V_1}$ resulta:

$$H(s) = \frac{s^2LCRa + bR}{s^2LCR + sL + R}$$

Dividiendo numerador y denominador por $LCR$:

$$\boxed{H(s) = \frac{s^2 a + \frac{b}{LC}}{s^2 + s\left(\frac{1}{RC}\right) + \frac{1}{LC}}}$$

Para un RLC paralelo:

$$\omega_0^2 = \frac{1}{LC} \qquad Q = \frac{R}{X_L} = \frac{R}{\omega_0 L} = R\,\omega_0 C$$

De donde:

$$\frac{\omega_0}{Q} = \frac{1}{RC} \quad (7) \qquad \text{y} \qquad \omega_0^2 = \frac{1}{LC} \quad (8)$$

Reemplazando (7) y (8) en $H(s)$:

$$H(s) = \frac{s^2 a + b\,\omega_0^2}{s^2 + s\left(\frac{\omega_0}{Q}\right) + \omega_0^2}$$

Con $\omega_0 = 1$, $R = 1\,\Omega$ y $Q = \frac{\sqrt{2}}{2}$:

$$\boxed{H(s) = \frac{a\,s^2 + b}{s^2 + \sqrt{2}\,s + 1}}$$

Y los componentes, para que opere en $f_0 = 10\ \text{kHz}$ con $R = 50\ \Omega$ (con el mismo $Q$), despejando de (7) y (8):

$$\omega_0 = 2\pi f_0 = 2\pi \cdot 10^4 \approx 62832\ \text{rad/s}$$

$$C = \frac{Q}{\omega_0\,R} \approx 225{,}1\ \text{nF} \qquad L = \frac{1}{\omega_0^2\,C} \approx 1{,}125\ \text{mH}$$

#### bonus.2) Respuesta de módulo y fase

La consigna no fija $a$ y $b$; tomo $a = b = 1$:

$$H(s) = \frac{s^2 + \omega_0^2}{s^2 + \sqrt{2}\,\omega_0\,s + \omega_0^2}$$

Reemplazo $s = j\omega$:

$$H(j\omega) = \frac{\omega_0^2 - \omega^2}{(\omega_0^2 - \omega^2) + j\sqrt{2}\,\omega_0\,\omega}$$

Módulo:

$$|H(j\omega)| = \frac{|\omega_0^2 - \omega^2|}{\sqrt{(\omega_0^2-\omega^2)^2 + 2\omega_0^2\omega^2}} = \frac{|\omega_0^2 - \omega^2|}{\sqrt{\omega_0^4 + \omega^4}}$$

Fase:

$$\varphi(\omega) = -\arctan\left(\frac{\sqrt{2}\,\omega_0\,\omega}{\omega_0^2 - \omega^2}\right)$$

(en $\omega_0$ el argumento pasa de $+\infty$ a $-\infty$, así que la fase salta de $-90°$ a $+90°$).

Asíntotas:

- $\omega \ll \omega_0$: $|H| \approx 1 \Rightarrow 0\ \text{dB}$, $\varphi \to 0°$
- $\omega \gg \omega_0$: $|H| \approx 1 \Rightarrow 0\ \text{dB}$, $\varphi \to 0°$
- En $\omega = \omega_0$ ($f_0 = 10\ \text{kHz}$): $|H| = 0 \Rightarrow -\infty\ \text{dB}$, la fase salta de $-90°$ a $+90°$


## Resultados

### Circuito a)

#### a.2) Análisis de la respuesta de módulo y fase

Lejos de $\omega_0$ la respuesta sigue a las asíntotas: $0\ \text{dB}$ en bajas frecuencias y $-40\ \text{dB/déc}$ en altas. Donde más se separa la curva real es en $\omega_0$, donde queda $3{,}01\ \text{dB}$ por debajo de las asíntotas. La fase pasa de $0°$ a $-180°$ y vale $-90°$ en $\omega_0$.

#### a.3) Tipo de filtro y ancho de banda

Es un **pasabajos de 2.º orden**: deja pasar las bajas frecuencias con ganancia $0\ \text{dB}$ y por encima de $\omega_0$ atenúa a $-40\ \text{dB/déc}$. Como $Q = \frac{\sqrt{2}}{2}$ no hay sobrepico (es la respuesta Butterworth, máximamente plana).

El corte de $-3\ \text{dB}$ sale de $|H(j\omega_c)|^2 = \frac{1}{1+\omega_c^4} = \frac{1}{2} \Rightarrow \omega_c = 1\ \text{rad/s}$, que coincide con $\omega_0$. La banda de paso va de $0$ a $\omega_c$:

$$BW = 1\ \text{rad/s}$$

### Circuito b)

#### b.2) Análisis de la respuesta de módulo y fase

En bajas frecuencias la respuesta sube a $+20\ \text{dB/déc}$ y en altas queda en $0\ \text{dB}$. Entre $\omega_0$ y $\omega_z = \sqrt{2}$ las asíntotas forman un pico de $+3{,}01\ \text{dB}$; la curva real queda por debajo ($|H(j\omega_0)| = \sqrt{3/2} \Rightarrow +1{,}76\ \text{dB}$), pero pasa por encima de $0\ \text{dB}$ por el cero en $\sqrt{2}$. La fase va de $+90°$ a $0°$.

#### b.3) Tipo de filtro y ancho de banda

Es un **pasaaltos de 2.º orden**: en continua la salida es nula ($H(0) = 0$) y en altas frecuencias $|H| \to 1$ ($0\ \text{dB}$). A diferencia de un pasaaltos de 2.º orden "puro", en bajas frecuencias atenúa solo $20\ \text{dB/déc}$ y no $40$, porque el numerador tiene un cero en el origen y otro en $-\sqrt{2}$ (no un cero doble en el origen).

El corte de $-3\ \text{dB}$ sale de $|H|^2 = \frac{1}{2}$:

$$\frac{\omega^2\,(2 + \omega^2)}{1 + \omega^4} = \frac{1}{2} \;\Rightarrow\; \omega^4 + 4\omega^2 - 1 = 0 \;\Rightarrow\; \omega_c^2 = \sqrt{5} - 2 \;\Rightarrow\; \omega_c = 0{,}486\ \text{rad/s}$$

La banda de paso va desde $\omega_c$ hasta infinito, así que no tiene un ancho de banda finito:

$$BW \to \infty \qquad (\omega > 0{,}486\ \text{rad/s})$$

### Circuito c)

#### c.2) Análisis de la respuesta de módulo y fase

La respuesta sube a $+20\ \text{dB/déc}$ hasta $\omega_0$ y baja a $-20\ \text{dB/déc}$ después. Las asíntotas se cortan en $+3{,}01\ \text{dB}$ y la curva real queda $3{,}01\ \text{dB}$ por debajo, en $0\ \text{dB}$. La fase pasa de $+90°$ a $-90°$ y vale $0°$ en $\omega_0$.

#### c.3) Tipo de filtro y ancho de banda

Es un **pasabanda de 2.º orden**: atenúa en continua y en altas frecuencias ($\pm 20\ \text{dB/déc}$) y tiene ganancia máxima de $0\ \text{dB}$ en $\omega_0 = 1\ \text{rad/s}$, donde la fase es $0°$.

Los cortes de $-3\ \text{dB}$ salen de $|H|^2 = \frac{1}{2}$:

$$\frac{2\omega^2}{1+\omega^4} = \frac{1}{2} \;\Rightarrow\; \omega^4 - 4\omega^2 + 1 = 0 \;\Rightarrow\; \omega_{1,2}^2 = 2 \mp \sqrt{3}$$

$$\omega_1 = 0{,}518\ \text{rad/s} \qquad \omega_2 = 1{,}932\ \text{rad/s}$$

$$BW = \omega_2 - \omega_1 = \frac{\omega_0}{Q} = \sqrt{2} \approx 1{,}414\ \text{rad/s}$$

### Bonus

#### bonus.2) Análisis de la respuesta de módulo y fase

Lejos de $f_0$ el módulo está en $0\ \text{dB}$ y en $f_0 = 10\ \text{kHz}$ cae a cero ($-\infty\ \text{dB}$). La fase baja de $0°$ a $-90°$ al acercarse a $f_0$, salta a $+90°$ y vuelve a $0°$ en altas frecuencias.

#### bonus.3) Tipo de filtro y ancho de banda

Con $a = b = 1$ es un **rechazabanda (notch) de 2.º orden**: deja pasar las bajas y las altas frecuencias con $0\ \text{dB}$ y anula la salida en $f_0 = 10\ \text{kHz}$, donde la fase salta de $-90°$ a $+90°$.

Los cortes de $-3\ \text{dB}$ salen de $|H|^2 = \frac{1}{2}$:

$$\frac{(\omega_0^2 - \omega^2)^2}{\omega_0^4 + \omega^4} = \frac{1}{2} \;\Rightarrow\; \omega^4 - 4\omega_0^2\,\omega^2 + \omega_0^4 = 0 \;\Rightarrow\; \omega_{1,2}^2 = (2 \mp \sqrt{3})\,\omega_0^2$$

$$f_1 = 0{,}518\,f_0 = 5{,}18\ \text{kHz} \qquad f_2 = 1{,}932\,f_0 = 19{,}32\ \text{kHz}$$

$$BW = f_2 - f_1 = \frac{f_0}{Q} = \sqrt{2} \cdot 10\ \text{kHz} \approx 14{,}14\ \text{kHz}$$

## Conclusiones

- Los cuatro circuitos tienen el mismo denominador, $s^2 + \frac{\omega_0}{Q}\,s + \omega_0^2$, o sea los mismos polos. Lo que cambia el tipo de filtro son los ceros, que dependen de cómo están conectados los elementos y de dónde se toma la salida: pasabajos (a), pasaaltos (b), pasabanda (c) y rechazabanda (bonus con $a = b = 1$).
- Con $Q = \frac{\sqrt{2}}{2}$, en a y c la curva real queda $3{,}01\ \text{dB}$ por debajo de las asíntotas en $\omega_0$. En el pasabajos el corte cae justo en $\omega_0$.
- En b el corte ($0{,}486\ \text{rad/s}$) no coincide con $\omega_0$ por el cero en $-\sqrt{2}$.
- En el pasabanda y en el rechazabanda el ancho de banda es $\frac{\omega_0}{Q} = \sqrt{2}\,\omega_0$.
- En el bonus, con $C = \frac{Q}{\omega_0 R}$ y $L = \frac{1}{\omega_0^2 C}$ el filtro opera en $f_0 = 10\ \text{kHz}$ con $R = 50\ \Omega$ y la respuesta mantiene la misma forma, corrida en frecuencia.
